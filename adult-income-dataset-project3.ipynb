## 7.Perform Data Cleaning [ Replace '?' with NaN ]

```python
import numpy as np
data['workclass']=data['workclass'].replace('?',np.nan)
data['occupation']=data['occupation'].replace('?',np.nan)
data['native-country']=data['native-country'].replace('?',np.nan)
data.isin(['?']).sum()
data.isnull().sum()
sns.heatmap(data.isnull())
```

## 8. Drop all The Missing Values

```python
per_missing = data.isnull().sum()*100/len(data)
data.dropna(how='any',inplace=True)
data.shape
```

## 9. Check For Duplicate Data and Drop Them

```python
dup = data.duplicated().any()
print("Are there any duplicated values in data:", dup)
data=data.drop_duplicates()
data.shape
```

## 17. Replace Salary Values With 0 and 1

```python
# the first option
data['income'].unique()
data['income'].value_counts()
sns.countplot('income', data=data)
```

```python
# the second option
def salary_data(sal):
    if sal == '<=50':
        return 0
    else:
        return 1
data['encoded_salary']=data['income'].apply(salary_data)
data.head(1)
```

```python
# the third option
data.replace(to_replace=['<=50K', '>50K'], value=[0,1], inplace=True)
data.head(1)
```
